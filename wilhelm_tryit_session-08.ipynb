## Try it yourself
----
**Key idea.** Each exercise makes one of this page's failure modes happen on purpose, which is the fastest way to
learn to recognise it.
1. Modify the gradient descent loop above to use a learning rate of 0.5, then 0.001. Describe (one sentence each) what happens to the loss and why.
2. Print the step number and current MSE every 100 iterations. Re-run at each learning rate above; inspect the gradient norm as well as the loss to distinguish convergence, slow progress, and divergence.
3. Re-run the first fit with `default_rng(0)` replaced by `default_rng(1)` , then with `default_rng(0)`
again. Confirm the third run reproduces the first exactly, and say which digits of `[5.09 2.95]` survive a change of seed.
4. Build a design matrix whose second column is the first plus `1e-9` times noise, then fit it with
`np.linalg.lstsq` and again with an explicit `np.linalg.inv(X.T @ X) @ X.T @ y` . Compare the
coefficients, then compare the two fits' MSE. Neither call raises an error or a warning: near-collinear is not singular, so `inv` hands back numbers as confidently as ever. In a typical run its predictions, not just its coefficients, come out visibly wrong, while `lstsq` 's wild-looking coefficients still predict well. The silence is the lesson.
