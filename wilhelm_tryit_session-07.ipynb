## Try it yourself
----
**Key idea.** The fastest way to trust the contract is to break it once, deliberately.
1. Make a corner pair plot of the penguins data coloured by species , and compute the between-group
over within-group spread ratio for each measurement, as in the iris example. Which pair separates the
species best?
2. Load iris with load_iris(return_X_y=True) , split 70/30 with stratify=y , and loop over
LogisticRegression , DecisionTreeClassifier , and KNeighborsClassifier . Change
random_state . Do the rankings move?
3. In the three-protocol experiment, move SelectKBest inside the pipeline but leave StandardScaler
outside. Predict the score before running it: which step was leaking?
4. Add a branch to the penguins ColumnTransformer routing body_mass_g through KBinsDiscretizer
instead of scaling. Does cross-validated accuracy change? Does the feature-name list?
