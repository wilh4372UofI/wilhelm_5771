## Try it yourself
----
**Key idea.** Each exercise turns one claim on this page into something you have measured.
1. (After Session 11.) Load the digits dataset ( `load_digits` , ten classes) and fit a multinomial logistic regression on scaled features. Print the 10 × 10 confusion matrix and find the two digits the model confuses most often in each direction. Then compare macro and weighted F1 and explain why they are
so close here but were far apart on the imbalanced three-class example.
2. On the breast cancer data, compute log loss for `LogisticRegression(C=100)` and
`LogisticRegression(C=0.01)` . Which has better accuracy, which has better cross-entropy, and what
does a disagreement between the two mean?
3. (After Session 11.) Take the three-blob dataset and shrink class 2 from 90 points to 30. Track macro F1, weighted F1, and the class-2 recall as it shrinks. Which number notices first?
