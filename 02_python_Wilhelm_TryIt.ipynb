```python
def argmax(values, key=None):
    largest_element_index = 0
    largest_element_value = key(values[0]) if key else values[0]

    for i in range(1, len(values)):
        current_value = key(values[i]) if key else values[i]

        if current_value > largest_element_value:
            largest_element_value = current_value
            largest_element_index = i

    return largest_element_index

number_elements = [3, -2, 7, 7, -7]

print(argmax(number_elements, key=abs))  #prints the index of the largest value.  expected 1 with homework elements.  Doesn't answer multiple indexes if elements are equal.
```

```python
class StandardScaler:
    """Standardize values using statistics learned from training data."""
    def fit(self, values):
        self.n_ = len(values)
        self.mean_ = sum(values) / self.n_
        self.min_ = min(values)
        self.range_ = max(values) - min(values)
        var = sum((v - self.mean_) ** 2 for v in values) / self.n_
        self.std_ = var ** 0.5
        return self # enables scaler.fit(...).transform(...)

    def transform(self, values):
        try: 
            return [(v - self.mean_) / self.std_ for v in values]
        except AttributeError:
            print("Error: Fit was not run before Transform.  self.mean_ and self.std_ don't exist yet.")
            raise SystemExit 

train_heights = [152.0, 168.0, 175.0, 181.0, 194.0]
test_heights = [160.0, 190.0]
scaler = StandardScaler()
scaled_train = scaler.fit(train_heights).transform(train_heights)
scaled_test = scaler.transform(test_heights) # reuses TRAINING mean/std
print(f"learned mean={scaler.mean_:.1f}, std={scaler.std_:.3f}, min={scaler.min_:.1f}, range={scaler.range_:.1f}")
# learned mean=174.0, std=13.928
print([round(v, 3) for v in scaled_test])
# [-1.005, 1.149]
```
