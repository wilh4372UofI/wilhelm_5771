```python
import seaborn as sns
import pandas as pd

tips = sns.load_dataset("tips")
tips["tip_pct"] = tips["tip"] / tips["total_bill"] * 100
print(tips.groupby("time", observed=True)["tip_pct"].mean().round(1))
```

```python
summary = tips.groupby("day", observed=True).agg(
    avg_tip=("tip", "mean"),
    avg_pct=("tip_pct", "mean"),
    n_parties=("tip", "size"),
    biggest_bill=("total_bill", "max"),
).round(2)
print(summary)
```

```python
print(tips.pivot_table(values="tip_pct", index="day", columns="time",
                       aggfunc="mean", margins=True, observed=True).round(1))
```

```python
summary = tips.groupby(["day", "time"], observed=True).agg(
    party_count=("tip", "size"),
    mean_tip_pct=("tip_pct", "mean"),
    max_bill=("total_bill", "max")
).round(2)
print(summary)
```

```python
# 2. Compute the Overall Pooled Survival Rate
overall_rate = titanic_raw["survived"].mean()

# 3. Compute rates and passenger counts for the 6 (pclass, sex) groups
# Using .agg() lets us see both the rate (mean) and the size (count) of each group
group_stats = (
    titanic_raw.groupby(["pclass", "sex"])["survived"]
    .agg(["mean", "count"])
    .reset_index()
)

# 4. Compute the Unweighted Mean of the 6 group rates
unweighted_mean = group_stats["mean"].mean()

# 5. Display the results
print(f"Overall Pooled Survival Rate: {overall_rate:.4f}")
print(f"Unweighted Mean of the 6 Groups: {unweighted_mean:.4f}\n")
print("--- Breakdown of the 6 Groups ---")
print(group_stats.to_string(index=False))
```
