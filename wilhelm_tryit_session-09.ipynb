## Try it yourself
----
**Key idea.** Each exercise makes one of this page's failure modes happen on purpose, which is the fastest way to learn to recognize it.
1. On the diabetes dataset ( `load_diabetes` ), fit `Lasso` with `alpha` in {0.01, 0.1, 1, 10} and plot the
number of nonzero coefficients versus `alpha` . Which features survive the longest? Look up what they
are.
2. Rerun the bias-variance experiment with `N_TRAIN = 200` instead of 40. Predict before you run it which
column moves and which stays put, then check.
3. Run `learning_curve` on a `KNeighborsClassifier` with `k = 1` and again with `k = 25` , on 800
points from `make_moons(noise=0.25)` . One curve keeps a gap that never closes and one converges; say
which is which before you plot, then check. Hint: first work out what each model's training score must
be, independent of sample size; the shape of the gap follows from that alone.
